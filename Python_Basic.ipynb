Python is a high-level, easy-to-learn programming language used for developing software, automating tasks, data analysis, machine learning, and many other applications.

Why is Python widely used in Data Analytics? — Three reasons
Easy to learn and use – Python has simple and readable syntax, making it beginner-friendly.
Powerful data analysis libraries – Libraries such as Pandas, NumPy, and Matplotlib make it easy to clean, analyze, and visualize data.
Large community and support – Python has a huge community, providing many tutorials, resources, and solutions to common data-analysis problems.

In short: Python is popular in Data Analytics because it is easy, powerful, and supported by many useful libraries and a large community.

Ans Q3--

A function in Python is a reusable block of code designed to perform a specific task. It can take inputs (arguments) and return an output.
Code Reusability – Write the code once and use it multiple times.
Reduces Repetition – Avoids writing the same code again and again.
Makes Code Organized – Breaks a large program into smaller, manageable parts.
Easy to Debug – Errors can be found and fixed more easily.

In short: A function helps make Python programs reusable, organized, and easier to maintain.
